## 8.2.2 PDF 로더

```python
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
import os

# 예제 파일 경로 설정 (실제 PDF 파일이 있어야 합니다.)
PDF_FILE_PATH = "한국은행 2024년 연차보고서.pdf"

if not os.path.exists(PDF_FILE_PATH):
    print(f"경고: {PDF_FILE_PATH} 파일이 없어 로딩을 건너뜁니다. 파일을 준비해주세요.")
else:
    # 1. 문서 로딩 (Document Loading)
    loader = PyMuPDFLoader(PDF_FILE_PATH)
    # .load() 메서드는 각 페이지를 하나의 Document 객체 리스트로 반환합니다.
    docs = loader.load()

    print(f"문서 로드 성공! 총 {len(docs)} 페이지가 로드되었습니다.")
    print(f"첫 페이지 메타데이터: {docs[0].metadata}")

    # 2. 문서 분할 (Text Splitting) 준비
    # RecursiveCharacterTextSplitter를 사용하여 청크를 나눕니다.
    # 청킹 크기와 오버랩은 LLM의 토큰 윈도우와 문서의 성격에 따라 조정이 필요합니다.
    splitter = RecursiveCharacterTextSplitter(
        chunk_size=1000,
        chunk_overlap=200 # 이전 청크와의 중첩을 통해 문맥 손실 방지
    )

    # 로드된 Document 객체 리스트를 청크로 분할
    chunks = splitter.split_documents(docs)

    print(f"청킹 완료. 총 {len(chunks)}개의 청크가 생성되었습니다.")
    print(f"첫 번째 청크 내용 일부:\n{chunks[0].page_content[:200]}...")
```

## 8.2.3 청킹 전략

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 텍스트 예시: 논리적 문단 구분과 문장 마침표가 포함됨
text = """
## 2.1 국내 기준금리 동향
2024년 상반기 한국은행은 기준금리를 3.5%로 유지했습니다. 이는 물가 안정 추세를 반영한 결과입니다.
하지만 미국 연준의 긴축 기조가 예상보다 강하게 이어질 경우, 하반기에는 금융 시장의 변동성이 확대될 수 있으며, 추가 인상 가능성이 제기됩니다.
"""

# 청킹 설정: 토큰 크기 100, 오버랩 20
splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    # 구분자 우선순위: 문단(두 줄바꿈) -> 문장(한 줄바꿈) -> 마침표 -> 공백
    separators=["\n\n", "\n", ".", " ", ""]
)

chunks = splitter.split_text(text)
print(f"--- RecursiveCharacterTextSplitter 실행 결과 (총 {len(chunks)}개 청크) ---")
for i, c in enumerate(chunks):
    print(f"\n[Chunk {i+1}] (길이: {len(c)})")
    print(c.strip())
```

```python
from docling.document_converter import DocumentConverter
from docling_core.transforms.chunker.hierarchical_chunker import HierarchicalChunker
from docling_core.transforms.chunker.hybrid_chunker import HybridChunker

# 문서 경로 설정 (실제 파일 경로로 수정 필요)
pdf_path = "/content/한국은행 2025년 9월중 금융시장 동향.pdf"

# 문서 변환
print("\n[1단계] 문서 변환 중...")
converter = DocumentConverter()
result = converter.convert(source=pdf_path)
doc = result.document

# 문서가 제대로 변환되었는지 확인
if doc is None:
    raise ValueError("문서 변환에 실패했습니다.")

print(f"문서 변환 완료. 페이지 수: {len(doc.pages) if hasattr(doc, 'pages') else 'N/A'}")

# 구조 기반 청킹
print("\n[2단계] 청킹 중...")
hc = HierarchicalChunker()
chunks = list(hc.chunk(dl_doc=doc))
print(f"청크 수: {len(chunks)}")

# 결과 출력 (선택사항)
print("\n=== 청크 샘플 (처음 3개) ===")
for i, chunk in enumerate(chunks[:3], 1):
    print(f"\n[청크 {i}]")
    print(f"텍스트: {chunk.text[:200]}...")  # 처음 200자만 출력
    if hasattr(chunk, 'meta'):
        print(f"메타데이터: {chunk.meta}")
```

## 8.2.4 벡터 데이터베이스

```python
from docling.document_converter import DocumentConverter
from langchain_openai import OpenAIEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document as LangChainDocument
import os

print("\n[3단계] LangChain Document 형식으로 변환 중...")

langchain_docs = []
for i, chunk in enumerate(chunks):
    # 메타데이터 추출
    metadata = {
        "chunk_id": i,
        "source": pdf_path,
        "char_count": len(chunk.text)
    }

    # 제목 정보 추출
    if hasattr(chunk, 'meta'):
        # 제목/헤딩 정보
        if hasattr(chunk.meta, 'headings') and chunk.meta.headings:
            headings = [h.text for h in chunk.meta.headings if hasattr(h, 'text')]
            metadata['section'] = " > ".join(headings) if headings else "본문"
        else:
            metadata['section'] = "본문"

        # 캡션 정보
        if hasattr(chunk.meta, 'captions') and chunk.meta.captions:
            captions = [c.text for c in chunk.meta.captions if hasattr(c, 'text')]
            if captions:
                metadata['caption'] = " | ".join(captions)

        # 문서 항목 타입
        if hasattr(chunk.meta, 'doc_items') and chunk.meta.doc_items:
            item_types = []
            for item in chunk.meta.doc_items:
                if hasattr(item, 'label'):
                    label = str(item.label).split('.')[-1]
                    item_types.append(label)
            if item_types:
                metadata['content_type'] = ', '.join(set(item_types))

        # 페이지 번호
        if hasattr(chunk.meta, 'doc_items') and chunk.meta.doc_items:
            for item in chunk.meta.doc_items:
                if hasattr(item, 'prov') and item.prov:
                    for prov in item.prov:
                        if hasattr(prov, 'page_no'):
                            metadata['page'] = prov.page_no
                            break
                    if 'page' in metadata:
                        break

    # LangChain Document 생성
    langchain_doc = LangChainDocument(
        page_content=chunk.text,
        metadata=metadata
    )
    langchain_docs.append(langchain_doc)

print(f"{len(langchain_docs)}개 LangChain Document 생성")

# 메타데이터 샘플 출력
print("\n메타데이터 샘플 (처음 3개):")
for i, doc in enumerate(langchain_docs[:3], 1):
    print(f"\n[Document {i}]")
    print(f"  섹션: {doc.metadata.get('section', 'N/A')}")
    print(f"  페이지: {doc.metadata.get('page', 'N/A')}")
    print(f"  타입: {doc.metadata.get('content_type', 'N/A')}")
    print(f"  텍스트: {doc.page_content[:100]}...")

# ========================================
# 3. OpenAI 임베딩 모델 설정
# ========================================
print("\n[4단계] OpenAI 임베딩 모델 설정 중...")

if "OPENAI_API_KEY" not in os.environ:
    raise ValueError("OPENAI_API_KEY 환경변수를 설정해주세요!")

# OpenAI 임베딩 모델
embedding_model = OpenAIEmbeddings(
    model="text-embedding-3-small",  # 또는 "text-embedding-3-large"
    # dimensions=1536,  # 차원 축소 가능 (선택사항)
)

print("OpenAI 임베딩 모델 설정 완료")
print("   모델: text-embedding-3-small")

# ========================================
# 4. 벡터 데이터베이스 생성
# ========================================
print("\n[5단계] Chroma 벡터 DB 생성 중...")

import re

def clean_text(text: str) -> str:
    # 이모지 및 ASCII 범위 밖 특수문자 제거
    return re.sub(r'[^\x00-\x7F\uAC00-\uD7A3\u4E00-\u9FFF\u3040-\u30FF]', '', text)

# langchain_docs 생성 후, Chroma 저장 전에 적용
for doc in langchain_docs:
    doc.page_content = clean_text(doc.page_content)
    doc.metadata = {k: clean_text(str(v)) for k, v in doc.metadata.items()}

db = Chroma.from_documents(
    documents=langchain_docs,
    embedding=embedding_model,
    collection_name="korean_financial_docs",
    persist_directory="./chroma_db"  # 영구 저장
)

print(f"벡터 DB 생성 완료: {len(langchain_docs)}개 문서 저장")

# ========================================
# 5. 검색 테스트
# ========================================
print("\n" + "=" * 70)
print("검색 테스트")
print("=" * 70)

test_queries = [
    "한국은행의 금리 인상 배경",
    "코스피 상승 이유",
    "9월 은행 대출 증가 현황"
]

for query in test_queries:
    print(f"\n질문: {query}")
    print("-" * 70)

    # 유사도 검색 (top 3)
    results = db.similarity_search(query, k=3)

    for i, result in enumerate(results, 1):
        print(f"\n[결과 {i}]")
        print(f"  섹션: {result.metadata.get('section', 'N/A')}")
        print(f"  페이지: {result.metadata.get('page', 'N/A')}")
        print(f"  타입: {result.metadata.get('content_type', 'N/A')}")
        print(f"  내용: {result.page_content[:200]}...")

    print("\n" + "=" * 70)

# ========================================
# 6. 고급 검색 (점수 포함)
# ========================================
print("\n유사도 점수 포함 검색")
print("=" * 70)

query = "금융시장 동향"
results_with_scores = db.similarity_search_with_score(query, k=3)

print(f"\n질문: {query}\n")
for i, (doc, score) in enumerate(results_with_scores, 1):
    print(f"[결과 {i}] 유사도: {1 - score:.4f}")  # 거리를 유사도로 변환
    print(f"  섹션: {doc.metadata.get('section', 'N/A')}")
    print(f"  내용: {doc.page_content[:150]}...")
    print()

print("=" * 70)
print("RAG 파이프라인 구축 완료!")
print("\n사용 방법:")
print("   db.similarity_search(query, k=3)  # 유사 문서 검색")
print("   db.similarity_search_with_score(query, k=3)  # 점수 포함")
print("   db.as_retriever(search_kwargs={'k': 5})  # Retriever로 사용")
```

## 8.3.1 Multi-vector retrieval

```python
# 1) 원본 텍스트용 DB
text_db = Chroma.from_documents(
    documents=langchain_docs,
    embedding=embedding_model,
    collection_name="text_embeddings",
    persist_directory="./chroma_db"
)
print(f"텍스트 DB: {len(langchain_docs)}개 저장")

# 2) 요약본용 DB
summary_docs = []
for doc in langchain_docs:
    # 요약 생성 (예시: 처음 200자)
    summary = doc.page_content[:200] + "..."
    summary_doc = LangChainDocument(
        page_content=summary,
        metadata=doc.metadata
    )
    summary_docs.append(summary_doc)

summary_db = Chroma.from_documents(
    documents=summary_docs,
    embedding=embedding_model,
    collection_name="summary_embeddings",
    persist_directory="./chroma_db"
)
print(f"요약 DB: {len(summary_docs)}개 저장")

# 3) 키워드용 DB
keyword_docs = []
for doc in langchain_docs:
    # 키워드 추출 (예시: 섹션명 + 타입)
    keywords = f"{doc.metadata.get('section', '')} {doc.metadata.get('content_type', '')}"
    keyword_doc = LangChainDocument(
        page_content=keywords,
        metadata=doc.metadata
    )
    keyword_docs.append(keyword_doc)

keyword_db = Chroma.from_documents(
    documents=keyword_docs,
    embedding=embedding_model,
    collection_name="keyword_embeddings",
    persist_directory="./chroma_db"
)
print(f"키워드 DB: {len(keyword_docs)}개 저장")

# ========================================
# EnsembleRetriever로 통합
# ========================================
from langchain_classic.retrievers import EnsembleRetriever

text_retriever = text_db.as_retriever(search_kwargs={"k": 5})
summary_retriever = summary_db.as_retriever(search_kwargs={"k": 5})
keyword_retriever = keyword_db.as_retriever(search_kwargs={"k": 5})

ensemble_retriever = EnsembleRetriever(
    retrievers=[text_retriever, summary_retriever, keyword_retriever],
    weights=[0.5, 0.3, 0.2]
)

print("Multi-vector RAG 파이프라인 완료!")
```

```python
# ========================================
# 검색 테스트
# ========================================
print("\n🔍 Multi-vector 검색 테스트")
query = "한국은행의 금리 인상 배경"
results = ensemble_retriever.invoke(query)

for i, result in enumerate(results[:3], 1):
    print(f"\n[결과 {i}]")
    print(f"  섹션: {result.metadata.get('section', 'N/A')}")
    print(f"  내용: {result.page_content[:150]}...")
```

## 8.3.3 Query construction

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# 1. LLM 초기화
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.0)

# 2. 쿼리 확장 프롬프트 정의
template = """
당신은 검색 쿼리 확장 전문가입니다.
주어진 사용자 질문을 분석하여, 검색 품질을 높일 수 있도록 다음 지침에 따라 3개의 독립적이고 상세한 검색 쿼리를 생성하세요.
지침:
1. 원본 질문의 의도를 최대한 유지하고, 다양한 관점과 관련 용어를 포함하세요.
2. 각 쿼리는 쉼표(,)로 구분하고, 다른 설명이나 텍스트 없이 쿼리 목록만 출력하세요.
사용자 질문: "{question}"
확장 쿼리 목록 (쉼표로 구분):
"""

prompt = ChatPromptTemplate.from_template(template)

# 3. LCEL 체인 구성 (프롬프트 -> LLM 호출 -> 문자열 파서)
expansion_chain = prompt | llm | StrOutputParser()

# 4. 쿼리 확장 실행
user_query = "코스피 상승 이유"
expanded_query = expansion_chain.invoke({"question": user_query})

# 결과 출력
print("확장된 쿼리 (검색엔진에 전달될 내용):")
expanded_queries = [q.strip() for q in expanded_query.split(',') if q.strip()]

for i, query in enumerate(expanded_queries, 1):
    print(f"  {i}. {query}")
```

## 8.3.4 HyDE

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.documents import Document as LangChainDocument

# HyDE를 위한 프롬프트 정의
hyde_template = """
다음 질문에 대해 검색 결과로 나타날 수 있는 이상적인 문서의 내용을 상상하여 작성해주세요.
이 문서는 질문에 대한 직접적인 답변을 포함하고 있어야 합니다.
질문: "{question}"
"""
hyde_prompt = ChatPromptTemplate.from_template(hyde_template)

# HyDE 체인 구성 (프롬프트 -> LLM 호출 -> 문자열 파서)
hyde_chain = hyde_prompt | llm | StrOutputParser()

# 가상 문서 생성
user_query = "코스피 상승 이유"
hypothetical_document_content = hyde_chain.invoke({"question": user_query})

print(f"생성된 가상 문서:\n{hypothetical_document_content[:200]}...\n")

# 가상 문서를 LangChain Document 객체로 변환
hypothetical_document = LangChainDocument(page_content=hypothetical_document_content)

# 가상 문서 임베딩
# 이전에 설정된 embedding_model을 사용합니다.
query_embedding = embedding_model.embed_documents([hypothetical_document.page_content])[0]

print(f"가상 문서 임베딩 생성 완료 (차원: {len(query_embedding)})")

# 생성된 임베딩을 사용하여 벡터 DB에서 검색 (선택 사항)
# db 객체가 이전 셀에서 생성되었다고 가정합니다.
if 'db' in locals():
    print("\n생성된 임베딩으로 벡터 DB 검색 테스트...")
    results = db.similarity_search_by_vector(query_embedding, k=3)

    for i, result in enumerate(results[:3], 1):
        print(f"\n[검색 결과 {i}]")
        print(f"  섹션: {result.metadata.get('section', 'N/A')}")
        print(f"  페이지: {result.metadata.get('page', 'N/A')}")
        print(f"  내용: {result.page_content[:150]}...")
else:
    print("\n'db' 벡터 데이터베이스가 정의되지 않아 검색 테스트를 건너뜁니다.")
```

## 8.5.5 비동기 검색 및 캐싱

```python
import asyncio
from langchain_core.prompts import ChatPromptTemplate

# 답변 생성을 위한 프롬프트 템플릿 정의
answer_template = """
주어진 문맥 정보를 사용하여 사용자의 질문에 답변하세요.
만약 문맥 정보만으로는 답변하기 어렵다면, 아는 정보 내에서 답변하거나 모른다고 말할 수 있습니다.
불필요하거나 관련 없는 정보는 포함하지 마세요.

문맥 정보:
{context}

사용자 질문:
{question}

답변:
"""
answer_prompt = ChatPromptTemplate.from_template(answer_template)


async def run_rag(query):
    # 1. 문서 검색 (비동기)
    docs_task = retriever.ainvoke(query)

    # 검색과 동시에 다른 작업(예: LLM 초기 호출 등)을 할 수 있지만,
    # 여기서는 검색 결과를 기다린 후 LLM 호출을 진행합니다.
    docs = await docs_task

    # 검색된 문서 내용을 하나의 문자열로 합치기
    context = "\n\n".join([doc.page_content for doc in docs])

    # 2. LLM 호출 (비동기) - 검색 결과를 프롬프트에 포함
    # answer_prompt를 사용하여 문맥과 질문을 포함한 프롬프트 생성
    llm_chain = answer_prompt | llm
    response_task = llm_chain.ainvoke({"context": context, "question": query})

    response = await response_task

    # LLM 응답과 검색된 문서를 함께 반환
    return docs, response.content # LLM 응답 객체에서 내용만 반환

# 예시 실행 (필요하다면 아래 주석을 해제하고 실행하세요)
async def main():
    query = "코스피 상승 이유"
    retrieved_docs, answer = await run_rag(query)
    print(f"질문: {query}")
    print(f"답변: {answer}")
    print("\n--- 사용된 문서 ---")
    for i, doc in enumerate(retrieved_docs):
        print(f"문서 {i+1} (페이지: {doc.metadata.get('page', 'N/A')}): {doc.page_content[:100]}...")

if __name__ == "__main__":
    # retriever와 llm 객체가 이전에 정의되어 있어야 합니다.
    if 'retriever' in locals() and 'llm' in locals():
        await main()
    else:
        print("retriever 또는 llm 객체가 정의되지 않았습니다. 이전 셀을 실행해주세요.")
```
