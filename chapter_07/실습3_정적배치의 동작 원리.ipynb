```python
import time
```

```python
def static_batch_server(requests, batch_size):
    """정적 배치 서버를 시뮬레이션합니다."""

    batched_requests = []
    for i in range(0, len(requests), batch_size):
        batch = requests[i:i + batch_size]
        batched_requests.append(batch)

    print(f"총 {len(requests)}개의 요청을 {len(batched_requests)}개의 배치로 나눕니다.\n")

    for i, batch in enumerate(batched_requests):
        print(f"--- 배치 {i+1} 처리 시작 ---")

        # 1. 패딩: 배치의 모든 시퀀스를 가장 긴 시퀀스 길이에 맞춤
        max_len = 0
        for req in batch:
            if req['len'] > max_len:
                max_len = req['len']

        print(f"배치 내 가장 긴 요청의 길이: {max_len} 토큰")
        print("모든 요청을 이 길이에 맞춰 패딩합니다.")

        # 실제로는 여기서 (batch_size, max_len) 크기의 텐서를 모델에 전달

        # 2. 토큰 생성 시뮬레이션 (선두 차단 문제 확인)
        # 모든 요청이 max_len 만큼의 스텝을 함께 진행
        for step in range(max_len):
            print(f"  > 생성 스텝 {step + 1}/{max_len} 진행 중...")
            for req in batch:
                if step < req['len']:
                    req['output'] += f"토큰{step+1} "
                else:
                    # 이미 생성이 끝났지만, 다른 요청 때문에 대기하며 낭비 발생
                    req['wasted_steps'] += 1
            time.sleep(0.1) # 각 스텝의 연산 시간 시뮬레이션

        print(f"\n--- 배치 {i+1} 처리 완료 ---")
        for req in batch:
            print(f"요청 ID {req['id']}:")
            print(f"  - 실제 길이: {req['len']}")
            print(f"  - 생성된 결과: {req['output']}")
            print(f"  - 낭비된 스텝 수 (선두 차단): {req['wasted_steps']}")
        print("-" * 25)

# --- 시뮬레이션 실행 ---
# 다양한 길이의 요청 4개
requests_to_process = [
    {'id': 'A', 'len': 5, 'output': '', 'wasted_steps': 0},
    {'id': 'B', 'len': 12, 'output': '', 'wasted_steps': 0},
    {'id': 'C', 'len': 3, 'output': '', 'wasted_steps': 0},
    {'id': 'D', 'len': 8, 'output': '', 'wasted_steps': 0},
]

static_batch_server(requests_to_process, batch_size=4)
```
