# sglang 서버 호출 테스트
- openai 호환 방식

```python
from openai import OpenAI

# sglang 서버는 OpenAI 호환 REST 제공
client = OpenAI(base_url="http://127.0.0.1:30000/v1", api_key="EMPTY")

resp = client.chat.completions.create(
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "한 줄로 자기소개 해줘."}
    ],
    max_tokens=128,
    temperature=0.7,
)
print(resp.choices[0].message.content)
```

- sglang 방식

```python
from sglang import function, user, assistant, gen, set_default_backend, RuntimeEndpoint

# 로컬에서 띄운 sglang 서버에 연결
set_default_backend(RuntimeEndpoint("http://127.0.0.1:30000"))

@function
def mini_chat(s):
    # sglang의 상태(state) 객체 's'에 프롬프트를 추가합니다.
    s += user("한 줄로 sglang을 소개해줘.")
    # 'assistant' 역할로 'reply'라는 변수에 최대 64토큰까지 생성을 지시합니다.
    s += assistant(gen("reply", max_tokens=64))

out = mini_chat.run()
print(out["reply"])
```
